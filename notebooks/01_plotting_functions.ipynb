# 01 — Plotting functions

This notebook focuses on:
- creating a clean x-grid,
- evaluating functions safely,
- plotting one or many functions,
- dealing with domain restrictions (like `log(x)` and `sqrt(x)`).


```python
from __future__ import annotations

from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

from calculus_with_python.grids import make_grid, safe_eval
from calculus_with_python.examples import sin, cos, log, sqrt
from calculus_with_python.plotting import plot_function, plot_many

out_dir = Path('..') / 'assets' / 'figures'
out_dir.mkdir(parents=True, exist_ok=True)
```

## 1) Create a grid

A grid should be strictly increasing and dense enough.


```python
x = make_grid(-2.0 * np.pi, 2.0 * np.pi, 1200)
x[:5], x[-5:]
```

## 2) Plot one function

Plot `sin(x)` and save the figure.


```python
ax = plot_function(sin, x, title='sin(x)', label='sin(x)')
fig = ax.figure
fig.savefig(out_dir / 'nb_plot_sin.png', dpi=160, bbox_inches='tight')
plt.close(fig)
```

## 3) Plot multiple functions

Plot `sin(x)` and `cos(x)` together.


```python
ax = plot_many({'sin(x)': sin, 'cos(x)': cos}, x, title='sin(x) and cos(x)')
fig = ax.figure
fig.savefig(out_dir / 'nb_plot_sin_cos.png', dpi=160, bbox_inches='tight')
plt.close(fig)
```

## 4) Domain restrictions

Some functions only make sense on part of the real line.
- `log(x)` requires `x > 0`
- `sqrt(x)` requires `x >= 0`


```python
x_pos = make_grid(1e-3, 10.0, 1000)

fig, ax = plt.subplots()
ax.plot(x_pos, safe_eval(log, x_pos), label='log(x)')
ax.plot(x_pos, safe_eval(sqrt, x_pos), label='sqrt(x)')
ax.set_title('Functions with restricted domain')
ax.set_xlabel('x')
ax.grid(True)
ax.legend()

fig.savefig(out_dir / 'nb_plot_domain_restrictions.png', dpi=160, bbox_inches='tight')
plt.close(fig)
```
