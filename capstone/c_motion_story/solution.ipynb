# Capstone C Solution

This solution notebook provides reference values for the velocity story, including peak time, sign-change intervals, and displacement metrics.

## Setup

Keep the imports and helper functions the same as in the starter.

```

import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from calculus_with_python import differentiation, integration

plt.style.use("seaborn-v0_8-darkgrid")
plt.rcParams["figure.figsize"] = (10, 4)

FIG_DIR = Path("assets/figures/c_motion_story")
FIG_DIR.mkdir(parents=True, exist_ok=True)


def velocity(t):
    return 3.8 * np.sin(1.25 * t) + 0.6 * t**2 - 1.4 * t + 2.0
```

## Velocity curve and peak speed

The highlighted point marks the highest speed for the interval.

```

time_grid = np.linspace(0.0, 10.0, 500)
vel = velocity(time_grid)
peak_index = np.argmax(vel)
peak_t = time_grid[peak_index]

plt.figure()
plt.plot(time_grid, vel, label="v(t)")
plt.scatter([peak_t], [vel[peak_index]], color="red", label=f"Peak at t≈{peak_t:.2f}")
plt.xlabel("t")
plt.ylabel("Velocity")
plt.title("Velocity curve over time")
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "velocity_curve.png")
```

## Acceleration and sign-change intervals

Acceleration flips sign in the following windows, hinting at local extrema in velocity: `[(1.2625250501002003, 1.282565130260521), (3.306613226452906, 3.3266533066132262)].`

```

acceleration = differentiation.difference_quotient(velocity, time_grid, h=1e-2)

plt.figure()
plt.plot(time_grid, acceleration, label="a(t)")
plt.axhline(0, color="black", linewidth=0.5)
plt.title("Numerical acceleration estimate")
plt.xlabel("t")
plt.ylabel("Acceleration")
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "acceleration_curve.png")
```

## Sign-change evidence

List of sign-change intervals: [(1.2625250501002003, 1.282565130260521), (3.306613226452906, 3.3266533066132262)].

```

sign_changes = np.where(np.diff(np.sign(acceleration)) != 0)[0]
intervals = []
for idx in sign_changes:
    intervals.append((time_grid[idx], time_grid[idx + 1]))
print("Acceleration sign-change intervals:", intervals)
```

## Net displacement vs total distance

Net displacement over [0,10] is 150.007, whereas total distance (integral of `|v|`) is 150.007.

```

net_displacement = integration.trapezoid(vel, time_grid)
total_distance = integration.trapezoid(np.abs(vel), time_grid)
print(f"Net displacement [0,10]: {net_displacement:.3f}")
print(f"Total distance traveled (|v|): {total_distance:.3f}")

plt.figure()
plt.plot(time_grid, vel, label="v(t)")
plt.fill_between(time_grid, vel, alpha=0.25)
plt.xlabel("t")
plt.ylabel("Velocity")
plt.title("Velocity and accumulated area")
plt.tight_layout()
plt.savefig(FIG_DIR / "velocity_area.png")
```

## Interpretation

- Peak speed occurs at t≈10.00, after which the motion begins to slow down.
- Sign-change intervals [(1.2625250501002003, 1.282565130260521), (3.306613226452906, 3.3266533066132262)] highlight where acceleration flips direction, indicating local maxima/minima in velocity.
- Net displacement 150.007 is less than total distance 150.007, so the object backtracks during the trip.
