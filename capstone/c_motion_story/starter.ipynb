# Capstone C: Find extrema and compute accumulated change

A realistic motion story: a velocity curve describes how an object speeds up and slows down. Work through each section to estimate acceleration, sign changes, and total displacement.

## Setup

Import numpy, matplotlib, and the repo utilities. Define the velocity function and prepare the figure directory.

```

import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from calculus_with_python import differentiation, integration

plt.style.use("seaborn-v0_8-darkgrid")
plt.rcParams["figure.figsize"] = (10, 4)

FIG_DIR = Path("assets/figures/c_motion_story")
FIG_DIR.mkdir(parents=True, exist_ok=True)


def velocity(t):
    return 3.8 * np.sin(1.25 * t) + 0.6 * t**2 - 1.4 * t + 2.0
```

## Velocity curve

Plot `v(t)` so you can spot the peak speed visually. TODO: describe what the highlighted peak means for motion.

```

time_grid = np.linspace(0.0, 10.0, 500)
vel = velocity(time_grid)
peak_index = np.argmax(vel)
peak_t = time_grid[peak_index]

plt.figure()
plt.plot(time_grid, vel, label="v(t)")
plt.scatter([peak_t], [vel[peak_index]], color="red", label=f"Peak at t≈{peak_t:.2f}")
plt.xlabel("t")
plt.ylabel("Velocity")
plt.title("Velocity curve over time")
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "velocity_curve.png")
```

## Acceleration via numerical derivative

Use the repo finite-difference helper to get acceleration. TODO: explain how the sign of acceleration ties to speeding up or slowing down.

```

acceleration = differentiation.difference_quotient(velocity, time_grid, h=1e-2)

plt.figure()
plt.plot(time_grid, acceleration, label="a(t)")
plt.axhline(0, color="black", linewidth=0.5)
plt.title("Numerical acceleration estimate")
plt.xlabel("t")
plt.ylabel("Acceleration")
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "acceleration_curve.png")
```

## Acceleration sign changes

Find time intervals where acceleration crosses zero. TODO: connect these intervals with potential changes in direction or local extrema.

```

sign_changes = np.where(np.diff(np.sign(acceleration)) != 0)[0]
intervals = []
for idx in sign_changes:
    intervals.append((time_grid[idx], time_grid[idx + 1]))
print("Acceleration sign-change intervals:", intervals)
```

## Accumulated change and distance

Compute both the net displacement (integral of `v`) and the total distance (integral of `|v|`). TODO: interpret these numbers in terms of stops, reversals, and total journey length.

```

net_displacement = integration.trapezoid(vel, time_grid)
total_distance = integration.trapezoid(np.abs(vel), time_grid)
print(f"Net displacement [0,10]: {net_displacement:.3f}")
print(f"Total distance traveled (|v|): {total_distance:.3f}")

plt.figure()
plt.plot(time_grid, vel, label="v(t)")
plt.fill_between(time_grid, vel, alpha=0.25)
plt.xlabel("t")
plt.ylabel("Velocity")
plt.title("Velocity and accumulated area")
plt.tight_layout()
plt.savefig(FIG_DIR / "velocity_area.png")
```
